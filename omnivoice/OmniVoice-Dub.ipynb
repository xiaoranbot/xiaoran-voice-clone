```python
# 上传当日日记 markdown（从 ran-blog 仓库或本地取），再跑批量生成。
# 把 DATE 改成当日日期；DIARY 改成当日日记文件路径（上传后在左侧文件区可见）。
from google.colab import files
print("上传当日日记 markdown：")
uploaded = files.upload()
diary_path = list(uploaded.keys())[0]
print(f"已上传：{diary_path}")
```

```python
import datetime
DATE = datetime.date.today().strftime("%Y-%m-%d")
print(f"当日：{DATE}，日记：{diary_path}")
!cd omnivoice-dub/omnivoice && python3 evening_batch.py --date {DATE} --diary /content/{diary_path} --out-dir /content/out
```

```python
# 下载当日音频（WAV + MP3）与声音 prompt（首次生成后记得保存回仓库）
from google.colab import files
import glob
for f in sorted(glob.glob("/content/out/*")):
    print(f)
    files.download(f)
```
